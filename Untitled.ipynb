````python
from transformers import AutoModelForCausalLM, AutoTokenizer


MODEL_NAME = "Qwen/Qwen2.5-7B-Instruct"
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    padding_side="left"
)

tokenizer.pad_token = tokenizer.eos_token
````

````python
PROMPT = """
Ты — эксперт по классификации профессиональных специалистов.

Проанализируй описание исполнителя и определи, в какой профессиональной области он работает.

Важно:
- Классифицируй именно профессию и навыки исполнителя.
- Не учитывай сферу клиентов или заказчиков.
- Не делай выводы о качестве, опыте или надежности специалиста.
- Если у специалиста несколько навыков, выбери основное направление работы.
- Категория должна быть достаточно общей, чтобы объединять похожих специалистов.
- Все ответы должны быть на русском языке.

Верни JSON:

{{
  "category": "",
  "subcategory": "",
  "skills": []
}}

Где:

category:
- широкая профессиональная область (например: IT, дизайн, маркетинг, строительство)

subcategory:
- более узкая специализация внутри области (например: разработка сайтов, графический дизайн, SEO)

skills:
- 3-7 ключевых профессиональных навыков специалиста.

Не используй слишком узкие или уникальные названия.
Не создавай категории вроде "специалист по настройке конкретного приложения".
Объединяй похожие профессии.

Примеры:

Описание:
"Разрабатывает сайты на WordPress и Tilda, занимается настройкой CMS, интеграцией сервисов."

Ответ:
{{
  "category": "IT-разработка",
  "subcategory": "Веб-разработка",
  "skills": [
    "WordPress",
    "Tilda",
    "CMS",
    "создание сайтов"
  ]
}}


Описание:
"Создает логотипы, фирменный стиль, рекламные баннеры и презентации."

Ответ:
{{
  "category": "Дизайн",
  "subcategory": "Графический дизайн",
  "skills": [
    "логотипы",
    "брендинг",
    "баннеры",
    "презентации"
  ]
}}


Описание:
"Проводит юридические консультации, готовит договоры и проверяет документы."

Ответ:
{{
  "category": "Юридические услуги",
  "subcategory": "Договорное право",
  "skills": [
    "договоры",
    "юридические консультации",
    "анализ документов"
  ]
}}


Описание специалиста:

{text}

Верни только JSON без пояснений.
"""

def parse_json_answer(answer):

    # убрать markdown
    answer = answer.replace(
        "```json",
        ""
    ).replace(
        "```",
        ""
    ).strip()


    match = re.search(
        r"\{.*\}",
        answer,
        re.DOTALL
    )


    if match:
        try:
            return json.loads(
                match.group()
            )
        except:
            pass


    return {
        "category": None,
        "subcategory": None,
        "task_summary": answer
    }



def make_prompt(text):

    messages = [
        {
            "role": "user",
            "content": PROMPT.format(text=text)
        }
    ]


    return tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )
````

````python
def classify_batch(
    texts,
    batch_size=8
):

    results = []

    for i in range(0, len(texts), batch_size):

        batch_texts = texts[i:i+batch_size]

        prompts = [
            make_prompt(x)
            for x in batch_texts
        ]


        inputs = tokenizer(
            prompts,
            padding=True,
            truncation=True,
            max_length=2048,
            return_tensors="pt"
        ).to(model.device)


        with torch.no_grad():

            outputs = model.generate(
                **inputs,
                max_new_tokens=80,
                do_sample=False,
                repetition_penalty=1.1,
                pad_token_id=tokenizer.eos_token_id
            )


        generated = outputs[
            :,
            inputs.input_ids.shape[1]:
        ]


        answers = tokenizer.batch_decode(
            generated,
            skip_special_tokens=True
        )


        for ans in answers:

            result = parse_json_answer(ans)

            results.append(result)


        if i % 100 == 0:
            print(
                f"{min(i+batch_size,len(texts))}/{len(texts)}"
            )


    return pd.DataFrame(results)
````

````python
def classify_dataframe(
    df,
    text_col,
    id_col,
    batch_size=8
):

    texts = df[text_col].tolist()

    predictions = classify_batch(
        texts,
        batch_size=batch_size
    )

    result = pd.concat(
        [
            df[[id_col]].reset_index(drop=True),
            predictions.reset_index(drop=True)
        ],
        axis=1
    )

    return result
````
